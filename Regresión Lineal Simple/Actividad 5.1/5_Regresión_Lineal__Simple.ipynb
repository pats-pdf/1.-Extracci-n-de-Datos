```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
#En caso de no poder cargar el archivo con seaborns leerlo con pandas
df= pd.read_csv("Analisis GAC Full7agosto 2026(Leads Reales).csv", skiprows=2)
```

```python
df.head()
```

```python
df.columns
```

```python
#quitamos columnas y filas que no nos interesan
df1 = df.iloc[:29, :11]
```

```python
#identificamos los datos nulos
df1.isnull().sum()
```

```python
df1[['Año', 'Mes', 'Total', 'Ilocalizable', 'No Contesta', 'D. Incorrectos',
       'Duplicados', 'Efectivos', 'Abiertos', 'Ventas', 'Conversion ']] = df1[['Año', 'Mes', 'Total', 'Ilocalizable', 'No Contesta', 'D. Incorrectos',
       'Duplicados', 'Efectivos', 'Abiertos', 'Ventas', 'Conversion ']].fillna(0) 
df1.head(5)
```

```python
#reemplazamos las filas nulas en las columnas con el año corrspondiente
df1.loc[1:12, 'Año'] = 2024
df1.loc[12:24, 'Año'] = 2025
df1.loc[24:, 'Año'] = 2026
```

```python
df1.dtypes
```

```python
df1
```

```python
# Convertimos a texto, eliminamos espacios y unificamos la escritura
df1['Mes'] = df1['Mes'].astype(str).str.strip().str.capitalize()

# Revisamos cómo quedaron
df1['Mes'].unique()
```

```python
meses = {
    'Enero': 1,
    'Febrero': 2,
    'Marzo': 3,
    'Abril': 4,
    'Mayo': 5,
    'Junio': 6,
    'Julio': 7,
    'Agosto': 8,
    'Septiembre': 9,
    'Octubre': 10,
    'Noviembre': 11,
    'Diciembre': 12
}

df1['Mes'] = df1['Mes'].map(meses)
```

```python
df1[['Mes']]
```

```python
df1
```

```python
df1.dtypes
```

```python
df1
```

```python
#quitamos el porcentaje para volverlo un valor numérico
df1['Conversion '] = df1['Conversion '].str.rstrip('%').astype(float)
```

```python
df1
```

```python
#Quitar las comas en la columna Total, Ilocalizable, No Contesta, D. Incorrectos, Duplicados, Efectivos, Abiertos y Ventas
df1['Total'] = df1['Total'].str.replace(',', '').astype(float)
```

```python
#Quitar las comas en la columna Total, Ilocalizable, No Contesta, D. Incorrectos, Duplicados, Efectivos, Abiertos y Ventas
df1['Efectivos'] = df1['Efectivos'].str.replace(',', '').astype(float)
```

```python
df1
```

```python
#Pasar todas las columnas a tipo numerico
df1 = df1.apply(pd.to_numeric, errors='coerce')
```

```python
df1
```

```python
#elimiamos los nulos
df1 = df1.fillna(0, inplace=True)
```

```python
df1
```

```python
#Imprimimos el scatter plot entre la variable dependiente (total) e independiente (alcohol) 
#para observar el comportamiento en su dispersión 
from turtle import color
sns.scatterplot(x='Ventas', y='Total', color="purple", data=df1)
#sns.scatterplot(x='speeding', y='total', color="red", data=df)
```

```python
#Declaramos las variables dependientes e independientes para la regresión lineal
#Vars_Indep= df[['alcohol', 'speeding']]
Vars_Indep= df1[['Ventas']]
Var_Dep= df1['Total']
```

```python
%pip install scikit-learn
```

```python
#Se define model como la función de regresión lineal
from sklearn.linear_model import LinearRegression
model= LinearRegression()
```

```python
#Ajustamos el modelo con las variables antes declaradas
model.fit(X=Vars_Indep, y=Var_Dep)
```

```python
#Verificamos los coeficientes obtenidos para el modelo ajustado
model.__dict__
```

```python
#Evaluamos la eficiencia del modelo obtenido por medio del coeficiente R Determinación
model.score(Vars_Indep,Var_Dep)
```

```python
#Predecimos los valores de total de accidentes a partir de la variable "alcohol"
#y_pred= model.predict(X=df[['alcohol', 'speeding']])
y_pred= model.predict(X=df1[['Ventas']])
y_pred
```

```python
#Insertamos la columna de predicciones en el DataFrame
df1.insert(0, 'Predicciones', y_pred)
df1
```

```python
#Visualizamos la gráfica comparativa entre el total real y el total predecido

sns.scatterplot(x='Ventas', y='Total', color="blue", data=df1)
sns.scatterplot(x='Ventas', y='Predicciones', color="red", data=df1)
sns.lineplot(x='Ventas', y='Predicciones', color="red", data=df1)
```

```python
#Corroboramos cual es el coeficiente de Determinación de nuestro modelo
coef_Deter=model.score(X=Vars_Indep, y=Var_Dep)
coef_Deter
```

```python
#Corroboramos cual es el coeficiente de Correlación de nuestro modelo
coef_Correl=np.sqrt(coef_Deter)
coef_Correl
```

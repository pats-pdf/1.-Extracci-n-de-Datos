```python
#Importamos las librerias pandas, numpy y matplotlib respectivamente
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

**VALORES NULOS**

```python
#Cargar archivo csv desde COLAB
#from google.colab import files
#files.upload()
```

```python
#Carga desde un archivo .xlsx sin indice
data = pd.read_excel('Base_de_Tlaxcala.xlsx')
data.head(5)
```

```python
#Eliminar columnas innecesarias
#axis = 1 para columnas
#axis = 0 para filas
data1= data.drop(['Correo','Telefono'], axis=1)
data1
```

```python
#corroborar valores nulos
valores_nulos=data1.isnull().sum()
valores_nulos
```

```python
#Identificar valores nulos por dataframe
valores_nulos=data1.isnull().sum().sum()
valores_nulos
```

**Métodos de Sustitución de Valores Nulos**

```python
#empezamos con "BDC" con ffill para llenar los valores nulos con el valor anterior, de esta manera podemos identificar los valores nulos y no perder información
data2=data1.copy()
data2["Producto"] = data2["Producto"].ffill()
```

```python
#corroborar valores nulos
valores_nulos=data2.isnull().sum()
valores_nulos
```

**Comparamos DataFrame3(Sustitución) y Dataframe4(Eliminación)**

**Convertir DataSet sin valores nulos a CSV**

```python
#Convertir DataFrame a CSV
data2.to_csv("Base_de_Tlaxcala_Sin_Nulos.csv", index=False)
```

**Métodos de Sustitución de Valores Nulos**

```python
#empezamos con "lead prom x mes" con .fillna para llenar con un valor especifico, en este caso "0" ya que representa ninguna venta
data1["Leads Prom X Mes"] =data1["Leads Prom X Mes"].fillna("0") 
```

```python
#corroborar valores nulos
valores_nulos=data1.isnull().sum()
valores_nulos
```

```python
#volvemos a ultilizar .fillna para llenar con un valor especifico, en este caso "0" ya que representa ninguna venta en piso
data1["T. Ventas Piso "] =data1["T. Ventas Piso "].fillna("0") 
```

```python
#corroborar valores nulos
valores_nulos=data1.isnull().sum()
valores_nulos
```

```python
#sustituimos nuevamente con 0 porque no aplica a los asesores que no tienen ventas en piso, de esta manera podemos identificar los valores nulos y no perder información
data1["Factor   "] =data1["Factor   "].fillna("0") 
```

```python
#corroborar valores nulos
valores_nulos=data1.isnull().sum()
valores_nulos
```

```python
data1["Total de Ventas Historicas "] =data1["Total de Ventas Historicas "].fillna("0") 
```

```python
#corroborar valores nulos
valores_nulos=data1.isnull().sum()
valores_nulos
```

```python
#sustituimos con 0 para las columnas de ventas historicas, raking ventas, dependencia de piso, historico de PDM, historico de SDC, factor PDM, factor SDC, raking PDM, raking SDC, conversion PDM y conversion SDC porque no aplica a los asesores que no tienen ventas en piso, de esta manera podemos identificar los valores nulos y no perder información
data1[["Total de Ventas Historicas ","RAKING VENTAS","Dependencia de Piso ","Historico de PDM","Historico de SDC","Factor  PDM","Factor  SDC","Raking PDM","Raking SDC","Conversion PDM","Conversion SDC"]] =data1[["Total de Ventas Historicas ","RAKING VENTAS","Dependencia de Piso ","Historico de PDM","Historico de SDC","Factor  PDM","Factor  SDC","Raking PDM","Raking SDC","Conversion PDM","Conversion SDC"]].fillna("0") 
```

```python
#corroborar valores nulos
valores_nulos=data1.isnull().sum()
valores_nulos
```

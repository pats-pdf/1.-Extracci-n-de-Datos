```python
#Carga desde un archivo .csv sin indice
data = pd.read_csv('Citas_Digital_Filtrado.csv')
data.head(5)
```

```python
#Eliminar columnas innecesarias
#axis = 1 para columnas
#axis = 0 para filas
data1= data.drop(['Unnamed: 15'], axis=1)
data1
```

```python
#Identificar valores nulos por columna
valores_nulos=data1.isnull().sum()
valores_nulos
```

```python
#Identificar valores nulos por dataframe
valores_nulos=data1.isnull().sum().sum()
valores_nulos
```

**Métodos de Sustitución de Valores Nulos**

```python
#empezamos con "BDC" con ffill para llenar los valores nulos con el valor anterior, de esta manera podemos identificar los valores nulos y no perder información
data2=data1.copy()
data2["BDC "] = data2["BDC "].ffill()
```

```python
#corroborar valores nulos
valores_nulos=data2.isnull().sum()
valores_nulos
```

```python
#a continuación sustituiremos "Fecha" con .fillna para llenar con un valor especifico, en este caso "2030-06-19 00:00:00" ya que es un valor que no se encuentra en el dataset y no afectará a los datos, de esta manera podemos identificar los valores nulos y no perder información

data2["Fecha"] =data2["Fecha"].fillna("2030-06-19 00:00:00") 
```

```python
#corroborar valores nulos
valores_nulos=data2.isnull().sum()
valores_nulos
```

```python
#a continuación sustituiremos "Nombre Cliente" con .fillna para llenar con un valor especifico, en este caso "Cliente no identificado" ya que es un valor que no se encuentra en el dataset y no afectará a los datos, de esta manera podemos identificar los valores nulos y no perder información
data2["Nombre Cliente "] =data2["Nombre Cliente "].fillna("Cliente no identificado") 
```

```python
#corroborar valores nulos
valores_nulos=data2.isnull().sum()
valores_nulos
```

```python
#a continuación sustituiremos "Telefono" con .fillna para llenar con un valor especifico, en este caso "Telefono no identificado" ya que es un valor que no se encuentra en el dataset y no afectará a los datos, de esta manera podemos identificar los valores nulos y no perder información
data2["Telefono "] =data2["Telefono "].fillna("Telefono no identificado")
```

```python
#corroborar valores nulos
valores_nulos=data2.isnull().sum()
valores_nulos
```

```python
#luego para la columna de "Estatus de Lead" vamos a sustituir los valores nulos con "Estatus no identificado" ya que es un valor que no se encuentra en el dataset y no afectará a los datos, de esta manera podemos identificar los valores nulos y no perder información
data2["Estatus de Lead"] =data2["Estatus de Lead"].fillna("Estatus no identificado")
```

```python
#corroborar valores nulos
valores_nulos=data2.isnull().sum()
valores_nulos
```

```python
#convertimos la columande Potencial de compra a tipo float para poder hacer operaciones matematicas con ella, primero eliminamos los caracteres que no son numeros y luego convertimos a float
data2['Potencial de compra'].unique

data2['Potencial de compra']=data2['Potencial de compra'].astype("string")
data2['Potencial de compra']=data2['Potencial de compra'].str.replace("Ventas ","0", regex=False)
data2['Potencial de compra']=data2['Potencial de compra'].str.replace("%","", regex=False)
data2['Potencial de compra']=data2['Potencial de compra'].str.replace("'","", regex=False)
data2['Potencial de compra']=data2['Potencial de compra'].str.replace("&","", regex=False)

data2['Potencial de compra']=data2['Potencial de compra'].astype(float)
```

```python
#seguimos con "Potencial de compra" con la mediana para llenar los valores nulos con el valor de la mediana, de esta manera podemos identificar los valores nulos y no perder información
data2["Potencial de compra"] = data2["Potencial de compra"].fillna(round(data2["Potencial de compra"].median(),1))
```

```python
#corroborar valores nulos
valores_nulos=data2.isnull().sum()
valores_nulos
```

```python
#a continuación vamos a sustituir "Venta" con .fillna para llenar con un valor especifico, en este caso "2" ya que es un valor que no se encuentra en el dataset y no afectará a los datos, de esta manera podemos identificar los valores nulos y no perder información
data2["Venta"] =data2["Venta"].fillna("2")
```

```python
#corroborar valores nulos
valores_nulos=data2.isnull().sum()
valores_nulos
```

```python
#para "asesor asignado" vamos a sustituir los valores nulos con "Asesor no identificado" ya que es un valor que no se encuentra en el dataset y no afectará a los datos, de esta manera podemos identificar los valores nulos y no perder información
data2["Asesor Asignado"] =data2["Asesor Asignado"].fillna("Asesor no identificado")
```

```python
#corroborar valores nulos
valores_nulos=data2.isnull().sum()
valores_nulos
```

```python
#para las preguntas despues de asesor asignado, utlizaremos el método fillna con el valor "No hay información" ya que es un valor que no se encuentra en el dataset y no afectará a los datos, de esta manera podemos identificar los valores nulos y no perder información
data2[["¿Por que no ha visitado la agencia?","¿Por que no hay solicitud de credito?","¿Por que no hubo prueba de manejo ?","¿Por que no hubo proceso wow?","¿Por que se asigno antes de vistar piso?"]] =data2[["¿Por que no ha visitado la agencia?","¿Por que no hay solicitud de credito?","¿Por que no hubo prueba de manejo ?","¿Por que no hubo proceso wow?","¿Por que se asigno antes de vistar piso?"]].fillna("No hay información")
```

```python
#Carga desde un archivo .xlsx sin indice
data = pd.read_excel('Análisis_de_Piso.xlsx')
data.head(5)
```

```python
#borre las primeras dos filas del archivo porque no tenían información relevante, esto para evitar nulos no relevantes en el análisis
data1= pd.read_excel('Análisis_de_Piso.xlsx', skiprows=2)
data1
```

```python
#Identificar valores nulos por columna
valores_nulos=data1.isnull().sum()
valores_nulos
```

```python
#Identificar valores nulos por dataframe
valores_nulos=data1.isnull().sum().sum()
valores_nulos
```

**Métodos de Sustitución de Valores Nulos**

```python
#a continuación sustituiremos "asesor" con .fillna para llenar con un valor especifico, en este caso "Asesor no identificado" ya que es un valor que no se encuentra en el dataset y no afectará a los datos, de esta manera podemos identificar los valores nulos y no perder información

data1["Asesor "] =data1["Asesor "].fillna("Asesor no identificado") 
```

```python
#corroborar valores nulos
valores_nulos=data1.isnull().sum()
valores_nulos
```

```python
a continuación sustituimos los valores nulos de las columnas de meses con .fillna para llenar con un valor especifico, en este caso "0" ya que no representa ninguna venta para los asesores que previamente limpiamos
data1[["feb","mar","abr","may","jun","jul","ago","sep","oct","nov","dic","ene","feb.1","mar.1","abr.1"]] =data1[["feb","mar","abr","may","jun","jul","ago","sep","oct","nov","dic","ene","feb.1","mar.1","abr.1"]].fillna("0") 
```

```python
#corroborar valores nulos
valores_nulos=data1.isnull().sum()
valores_nulos
```

```python
#repetimos los mismos pasos para las columnas de meses que se encuentran al final del dataset, ya que también contienen valores nulos, sustituyendo con 0
data1[["Unnamed: 18","Unnamed: 19","mar.2","abr.2","may.1","jun.1","jul.1","ago.1","sep.1","oct.1","nov.1","dic.1","ene.1","feb.3","mar.3","May ","Junio"]] =data1[["Unnamed: 18","Unnamed: 19","mar.2","abr.2","may.1","jun.1","jul.1","ago.1","sep.1","oct.1","nov.1","dic.1","ene.1","feb.3","mar.3","May ","Junio"]].fillna("0") 
```

```python
#corroborar valores nulos
valores_nulos=data1.isnull().sum()
valores_nulos
```

# ACTIVIDAD 1.1 - CITAS DIGITAL - A01730600

```python
#Cargar el archivo
df = pd.read_excel('Citas_Digital.xlsx')
df
```

```python
#Paso 1. Filtro por filas (0-673)
filtro_filas = df.iloc[0:673,: ]
filtro_filas.info()
```

```python
#Paso 2. Filtro por columnas (0-2, 4-14)
filtro_columnas = filtro_filas.iloc[:, [0,1,2,4,5,6,7,8,9]]
filtro_columnas.info()
```

```python
#Paso 3.Convertir los datos de la variable “Potencial de compra” a tipo float
filtro_columnas['Potencial de compra'].unique

filtro_columnas['Potencial de compra']=filtro_columnas['Potencial de compra'].astype("string")
filtro_columnas['Potencial de compra']=filtro_columnas['Potencial de compra'].str.replace("Ventas ","0", regex=False)
filtro_columnas['Potencial de compra']=filtro_columnas['Potencial de compra'].str.replace("%","", regex=False)
filtro_columnas['Potencial de compra']=filtro_columnas['Potencial de compra'].str.replace("'","", regex=False)
filtro_columnas['Potencial de compra']=filtro_columnas['Potencial de compra'].str.replace("&","", regex=False)

filtro_columnas['Potencial de compra']=filtro_columnas['Potencial de compra'].astype(float)
```

```python
filtro_columnas.info()
```

```python
#Paso 4. Obtener el CSV General y nombrarlo “Citas_Digital_General.csv”
filtro_columnas.to_csv("Citas_Digital_General.csv", index=False)
```

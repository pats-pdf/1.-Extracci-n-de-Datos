```python
# Extraigo la pregunta abierta
Pregunta_abierta = bitacora_piso[
    ["¿Por que no hubo prueba de manejo?"]
].copy()

# Elimino las respuestas vacías
Pregunta_abierta = Pregunta_abierta.dropna()

Pregunta_abierta
```

```python
Pregunta_abierta["Caracteristica"] = "Otro"

Pregunta_abierta.loc[
    Pregunta_abierta["¿Por que no hubo prueba de manejo?"]
    .str.contains("tiempo", case=False),
    "Caracteristica"
] = "Falta de tiempo"

Pregunta_abierta.loc[
    Pregunta_abierta["¿Por que no hubo prueba de manejo?"]
    .str.contains("unidad", case=False),
    "Caracteristica"
] = "Unidad no disponible"

Pregunta_abierta.loc[
    Pregunta_abierta["¿Por que no hubo prueba de manejo?"]
    .str.contains("no desea|no quiso", case=False),
    "Caracteristica"
] = "Cliente no quiso"

Pregunta_abierta.loc[
    Pregunta_abierta["¿Por que no hubo prueba de manejo?"]
    .str.contains("sábado|acudir|domicilio", case=False),
    "Caracteristica"
] = "Prueba posterior"

Pregunta_abierta.loc[
    Pregunta_abierta["¿Por que no hubo prueba de manejo?"]
    .str.contains("sí hubo|sí realizo|ya.*prueba", case=False),
    "Caracteristica"
] = "Sí hubo prueba de manejo"

Pregunta_abierta
```

```python
Tabla_caracteristicas = (
    Pregunta_abierta["Caracteristica"]
    .value_counts()
    .reset_index()
)

Tabla_caracteristicas
```

```python
#Realizamos grafico de pastel del dataframe filtrado
ax = Tabla_caracteristicas["count"].plot(
    kind="pie",
    labels=Tabla_caracteristicas["Caracteristica"],
    figsize=(10, 5),
    shadow=True,
    autopct="%0.1f%%",
    legend=False
)

ax.set_ylabel("")
ax.set_title("Motivos relacionados con la prueba de manejo")
```

```python
#Extraigo la pregunta abierta
Pregunta_credito = bitacora_piso[
    ["¿Por que no hubo Solicitud de credito?"]
].copy()

# Elimino respuestas vacías
Pregunta_credito = Pregunta_credito.dropna()
```

```python
#Categoría inicial
Pregunta_credito["Caracteristica"] = "Otro"


# 3. Clasifico las respuestas
Pregunta_credito.loc[
    Pregunta_credito["¿Por que no hubo Solicitud de credito?"]
    .str.contains("contado", case=False),
    "Caracteristica"
] = "Compra de contado"

Pregunta_credito.loc[
    Pregunta_credito["¿Por que no hubo Solicitud de credito?"]
    .str.contains("hijos|decisión", case=False),
    "Caracteristica"
] = "Decisión familiar"

Pregunta_credito.loc[
    Pregunta_credito["¿Por que no hubo Solicitud de credito?"]
    .str.contains("compromisos económicos|capacidad de pago|historial", case=False),
    "Caracteristica"
] = "Falta de capacidad de pago"

Pregunta_credito.loc[
    Pregunta_credito["¿Por que no hubo Solicitud de credito?"]
    .str.contains("cooperativa", case=False),
    "Caracteristica"
] = "Crédito por otra institución"

Pregunta_credito.loc[
    Pregunta_credito["¿Por que no hubo Solicitud de credito?"]
    .str.contains("mamá|financiera", case=False),
    "Caracteristica"
] = "Dificultad para obtener crédito"

Pregunta_credito.loc[
    Pregunta_credito["¿Por que no hubo Solicitud de credito?"]
    .str.contains("no será pronto|valorando opciones|esperar|despues|después", case=False),
    "Caracteristica"
] = "Decisión aplazada"

Pregunta_credito.loc[
    Pregunta_credito["¿Por que no hubo Solicitud de credito?"]
    .str.contains("no quiere tramitar", case=False),
    "Caracteristica"
] = "Cliente no quiso"

# Esta categoría va al final porque confirma que sí existió solicitud
Pregunta_credito.loc[
    Pregunta_credito["¿Por que no hubo Solicitud de credito?"]
    .str.contains("sí hubo SDC|si hubo SDC|ingresa SDC", case=False),
    "Caracteristica"
] = "Sí hubo solicitud de crédito"
```

```python
Tabla_credito = (
    Pregunta_credito["Caracteristica"]
    .value_counts()
    .reset_index()
)

Tabla_credito
```

```python
Tabla_credito_sin = Tabla_credito[
    Tabla_credito["Caracteristica"] != "Sí hubo solicitud de crédito"
]

Tabla_credito_sin
```

```python
ax = Tabla_credito_sin["count"].plot(
    kind="pie",
    labels=Tabla_credito_sin["Caracteristica"],
    figsize=(10, 5),
    shadow=True,
    autopct="%0.1f%%",
    legend=False
)

ax.set_ylabel("")
ax.set_title("Motivos por los que no hubo solicitud de crédito")
```

```python
#Extraemos la pregunta abierta
Pregunta_compra = bitacora_piso[
    ["¿Por que no se ha formalizado la compra?"]
].copy()

# Elimino respuestas vacías
Pregunta_compra = Pregunta_compra.dropna()
```

```python
#Creo una categoría inicial
Pregunta_compra["Caracteristica"] = "Otro"


#Clasifico las respuestas
Pregunta_compra.loc[
    Pregunta_compra["¿Por que no se ha formalizado la compra?"]
    .str.contains("próximo año|proximo año", case=False),
    "Caracteristica"
] = "Compra aplazada"

Pregunta_compra.loc[
    Pregunta_compra["¿Por que no se ha formalizado la compra?"]
    .str.contains("rechazado|no le fue otorgado|buró|buro", case=False),
    "Caracteristica"
] = "Crédito rechazado"

Pregunta_compra.loc[
    Pregunta_compra["¿Por que no se ha formalizado la compra?"]
    .str.contains("enganche|movimientos en sus cuentas|moviemientos", case=False),
    "Caracteristica"
] = "Falta de recursos"

Pregunta_compra.loc[
    Pregunta_compra["¿Por que no se ha formalizado la compra?"]
    .str.contains("no volvio|no volvió|no contestar", case=False),
    "Caracteristica"
] = "Cliente sin contacto"

Pregunta_compra.loc[
    Pregunta_compra["¿Por que no se ha formalizado la compra?"]
    .str.contains("valorando|primera vez", case=False),
    "Caracteristica"
] = "Evaluando opciones"

Pregunta_compra.loc[
    Pregunta_compra["¿Por que no se ha formalizado la compra?"]
    .str.contains("Tiguan", case=False),
    "Caracteristica"
] = "Prefiere otro vehículo"

Pregunta_compra.loc[
    Pregunta_compra["¿Por que no se ha formalizado la compra?"]
    .str.contains("inversion|inversión", case=False),
    "Caracteristica"
] = "Prioridad de inversión"

Pregunta_compra.loc[
    Pregunta_compra["¿Por que no se ha formalizado la compra?"]
    .str.contains("apartado", case=False),
    "Caracteristica"
] = "Compra en proceso"

# Estas respuestas confirman que la compra sí se formalizó
Pregunta_compra.loc[
    Pregunta_compra["¿Por que no se ha formalizado la compra?"]
    .str.contains("ya se formalizo|ya se formalizó|entrego la camioneta", case=False),
    "Caracteristica"
] = "Compra formalizada"
```

```python
# Creo la tabla de frecuencias
Tabla_compra = Pregunta_compra["Caracteristica"].value_counts().reset_index()
Tabla_compra
```

```python
# Gráfica de pastel
ax = Tabla_compra["count"].plot(
    kind="pie",
    labels=Tabla_compra["Caracteristica"],
    figsize=(10, 5),
    shadow=True,
    autopct="%0.1f%%",
    legend=False
)

ax.set_ylabel("")
ax.set_title("Razones relacionadas con la formalización de la compra")
```

```python
def analizar_sturges(dataframe, variable):

    # Elimino únicamente los nulos de la variable analizada

    datos = dataframe[variable].dropna()

    # Calculo el número de observaciones

    n = len(datos)

    # Aplico la regla de Sturges

    k = math.ceil(
        1 + 3.322 * math.log10(n)
    )

    # Obtengo el valor mínimo y máximo

    minimo = datos.min()
    maximo = datos.max()

    # Calculo la amplitud de los intervalos

    amplitud = (maximo - minimo) / k

    # Creo los límites de los intervalos

    limites = np.linspace(
        minimo,
        maximo,
        k + 1
    )

    # Clasifico los datos dentro de los intervalos

    categorias = pd.cut(
        datos,
        bins=limites,
        include_lowest=True
    )

    # Creo la tabla de frecuencias

    tabla_frecuencias = (
        categorias
        .value_counts(sort=False)
        .rename_axis("Intervalo")
        .reset_index(name="Frecuencia")
    )

    # Defino la cantidad de decimales de las etiquetas

    if variable in ["Conversion", "Efectividad de Leads"]:

        decimales = 2

    else:

        decimales = 0

    # Creo etiquetas más claras para los intervalos

    etiquetas = []

    for intervalo in tabla_frecuencias["Intervalo"]:

        limite_inferior = max(
            intervalo.left,
            minimo
        )

        limite_superior = min(
            intervalo.right,
            maximo
        )

        etiqueta = (
            f"{limite_inferior:.{decimales}f} "
            f"a {limite_superior:.{decimales}f}"
        )

        etiquetas.append(etiqueta)

    tabla_frecuencias["Rango"] = etiquetas

    # Creo la tabla que se mostrará

    tabla_final = tabla_frecuencias[
        ["Rango", "Frecuencia"]
    ].copy()

    # Muestro los resultados del cálculo

    print("=" * 50)
    print("Variable:", variable)
    print("Número de observaciones:", n)
    print("Número de intervalos:", k)
    print("Valor mínimo:", minimo)
    print("Valor máximo:", maximo)
    print("Amplitud:", round(amplitud, 2))
    print("=" * 50)

    display(tabla_final)

    plt.figure(figsize=(9, 5))

    barras = plt.bar(
        tabla_final["Rango"],
        tabla_final["Frecuencia"],
        color="firebrick",
        edgecolor="black"
    )

    # Coloco las frecuencias encima de las barras

    plt.bar_label(
        barras,
        padding=3
    )

    # Defino el título y la unidad de cada variable

    if variable == "Ventas":

        titulo = "Frecuencia por rangos de ventas"
        nombre_eje_x = "Rangos de ventas"

    elif variable in ["Conversion", "Efectividad de Leads"]:

        titulo = f"Frecuencia por rangos de {variable} (%)"
        nombre_eje_x = "Rangos de porcentaje (%)"

    else:

        titulo = f"Frecuencia por rangos de {variable} (leads)"
        nombre_eje_x = "Rangos de cantidad de leads"

    # Coloco título y nombres de los ejes

    plt.title(titulo)
    plt.xlabel(nombre_eje_x)
    plt.ylabel("Frecuencia")

    plt.xticks(
        rotation=45,
        ha="right"
    )

    plt.grid(
        axis="y",
        alpha=0.3
    )

    plt.tight_layout()
    plt.show()

    return tabla_final
```
